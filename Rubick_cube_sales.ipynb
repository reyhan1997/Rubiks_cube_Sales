```python
all_sales = pd.concat([sales_2023, sales_2024, sales_2025])
all_sales['Date'] = pd.to_datetime(all_sales['Date'])
all_sales['Revenue'] = all_sales['Units Sold'] * all_sales['Unit Price']
daily_revenue = all_sales.groupby('Date')['Revenue'].sum().reset_index()

plt.figure(figsize=(15, 7))
sns.lineplot(data=daily_revenue, x='Date', y='Revenue')
plt.title('Total Revenue Each Day (2023-2025)')
plt.xlabel('Date')
plt.ylabel('Total Revenue')
plt.ylim(0, 55000)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
sales_2023_daily = sales_2023.copy()
sales_2023_daily['Date'] = pd.to_datetime(sales_2023_daily['Date'])
sales_2023_daily['Revenue'] = sales_2023_daily['Units Sold'] * sales_2023_daily['Unit Price']

daily_summary_2023 = sales_2023_daily.groupby('Date').agg(
    Total_Revenue=('Revenue', 'sum'),
    Total_Units_Sold=('Units Sold', 'sum')
).reset_index()

print("Daily Total Revenue and Total Quantity for 2023:")
print(daily_summary_2023)
```

```python
sales_2023_daily['Revenue'] = sales_2023_daily['Units Sold'] * sales_2023_daily['Unit Price']

# Calculate daily revenue per product
daily_product_revenue_2023 = sales_2023_daily.groupby(['Date', 'Product Name'])['Revenue'].sum().reset_index()

plt.figure(figsize=(15, 8))
sns.lineplot(data=daily_product_revenue_2023, x='Date', y='Revenue', hue='Product Name', marker='o')

plt.title('Daily Revenue by Product Name for 2023')
plt.xlabel('Date')
plt.ylabel('Total Revenue')
plt.xticks(rotation=45, ha='right')
plt.legend(title='Product Name', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True)
plt.tight_layout()
plt.show()
```
